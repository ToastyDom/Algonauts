# **Overview**

## **The Algonauts Project**

The [Algonauts Project](https://algonautsproject.com/), first launched in 2019, is on a mission to bring biological and machine intelligence researchers together on a common platform to exchange ideas and explore the intelligence frontier. Inspired by astronauts’ exploration of space, “algonauts” explore biological and artificial intelligence with state-of-the-art algorithmic tools, advancing both fields.

Previous challenges studied brain responses to images, short videos and multimodal movies. The 2027 challenge extends this progression to active video game play.

## **The 2027 challenge**


Encoding models are used as predictive and explanatory tools in computational neuroscience. They take features describing sensory input, actions or task events and predict the corresponding neural responses, helping us study how these aspects of a task are represented in the brain.

The goal of the Algonauts Project 2027 Challenge is to develop functional magnetic resonance imaging (fMRI) encoding models that predict brain responses during video game play and that generalize to new levels and games. During gameplay, players’ decisions and actions influence what they subsequently see and hear (and vice versa). Predicting brain responses in this setting therefore requires models to account for the relationship between sensory input and player behavior.

The challenge is based on data from the **[Courtois NeuroMod Project (CNeuroMod)](https://www.cneuromod.ca/)**, in which subjects played four retro video games while their brain activity was recorded with fMRI. Challenge participants receive gameplay video, audio, controller actions and annotated game events to predict responses across **1,134 brain parcels**. Controller actions describe players’ inputs, while game events describe occurrences such as collecting items, taking damage or defeating enemies.

The challenge provides data from four retro video games: three from the Mario family and a more distinct action game from the Shinobi series. **Super Mario Bros. (NES, 1985)** is the first installment in the Super Mario series, a side-scrolling platform game in which players navigate obstacles, collect items and avoid or defeat enemies. The **Super Mario All-Stars (SNES, 1993)** recordings feature a visually and acoustically updated version of Super Mario Bros., retaining its original level layouts. **Super Mario Bros. 3 (NES, 1988)** introduces a different visual style, new environments and additional gameplay mechanics, including power-ups that allow Mario to fly.

Further from the Mario family, **Shinobi III: Return of the Ninja Master (Sega Genesis, 1993)** is an action platform game in which players control a ninja and fight enemies using throwing weapons, close-range attacks and special abilities. Its visual setting, combat mechanics and movement abilities provide a different gameplay environment from the Mario games. The table below summarizes the available data for each game.

| Dataset key | Game | Subjects | Unique levels | Gameplay |
|---|---|---|---:|---:|
| `mario` | Super Mario Bros. (NES) | 01, 02, 03, 05, 06 | 22 | ~84 h |
| `mariostars` | Super Mario All-Stars (SNES) | 01, 02, 03, 05, 06 | 22 | ~31 h |
| `mario3` | Super Mario Bros. 3 (NES) | 01, 02, 03, 05, 06 | 80 | ~41 h |
| `shinobi` | Shinobi (Sega Genesis) | 01, 02, 04, 06 | 3 | ~31 h |

Gameplay durations are totals across all available subjects. Subject coverage differs across games. This tutorial uses **`sub-01`** for section 1-2, who played all four games. All evaluation subejcts are used for sections 3 & 4. Gameplay recordings are subject-specific because each subject follows their own trajectory through the games.

**Before exploring the data, read the README within each game’s directory in the challenge repository.** It describes the game, its recording setup, the available data and any specific characteristics or caveats.

## **Challenge phases**

The challenge consists of a Development Phase and a Generalization Phase, followed by an indefinite Post Challenge Phase serving as a public benchmark. These evaluations test **in-distribution (ID) generalization** to held out levels from a game represented during training and **out-of-distribution (OOD) generalization** to games not represented during training.

1. **Development Phase:** challenge participants develop encoding models using released gameplay and corresponding fMRI responses from Super Mario Bros., Super Mario All-Stars, Super Mario Bros. 3 and Shinobi. Selected levels from **Super Mario Bros. 3** are held out for ID evaluation: their gameplay video, audio, controller actions and game events are released, while the corresponding fMRI responses remain hidden.

   Participants submit predicted fMRI responses to CodaBench and receive prediction accuracy scores through a public leaderboard. This feedback allows them to assess and improve ID generalization to held out levels.

2. **Generalization Phase:** challenge participants predict fMRI responses to games not represented in the Development Phase data, evaluating OOD generalization. The identities of these games remain hidden until this phase begins. Approximately two hours of gameplay per subject are released, including video, audio, controller actions and game events, while the corresponding fMRI responses remain hidden.

   Participants submit predictions to CodaBench for OOD evaluation. **Performance in this phase determines the final challenge ranking.** Note that models that perform similarly on held out Mario 3 levels may differ in their OOD generalization, meaning how well they transfer to these new gameplay environments.

3. **Post Challenge Phase:** following the Development and Generalization Phases, the challenge remains open indefinitely as a public benchmark. Researchers can continue uploading their models and obtain separate scores for ID generalization and OOD generalization.

Prediction accuracy is measured using **Pearson’s correlation coefficient, r**. For each evaluated subject and game, predicted and measured responses are concatenated across all valid time points from the evaluated recordings. A correlation is calculated separately for each of the 1,134 brain parcels. Parcel correlations are averaged to obtain a score for each subject and game. Subject scores are then averaged within each game. If several games are evaluated, their scores are averaged with equal weight.

The challenge evaluates `sub-01`, `sub-02`, `sub-03`, `sub-05` and `sub-06`. Available training data differ across subjects because not every subject played every game. Subject 4 only played Shinobi and is excluded from challenge evaluation, but its recordings are retained as additional training data that challenge participants may use to improve their models.

### **1.1.2 | Event annotations**

Gameplay recordings provide video and audio. The accompanying `.tsv` files provide timestamped annotations for two further stimulus types:

* **Controller actions** describe the player’s button presses, such as moving right or jumping. Their labels begin with `Action/`, for example `Action/Right` or `Action/Jump`. Their durations indicate how long the buttons were held.
* **Game events** describe occurrences during gameplay, such as collecting an item, taking damage or completing a level. Examples include `Item_collected/Coin`, `Player_damaged` and `Level_completed`. Some events are instantaneous; others describe sustained states.

These annotations are stored alongside the gameplay recordings and follow this naming convention:

```text
sub-<subject>_ses-<session>_task-<game>_run-<run>_desc-annotated_events.tsv
```

**One events table covers an entire fMRI run, which can contain several gameplay recordings.** Controller actions and game events use timestamps measured from the beginning of that run.

To locate each gameplay recording on the same timeline, the table also includes rows labeled `gym-retro_game`. These rows identify the recording’s file, starting time and duration. They provide the timing information needed to match the audiovisual recordings with their annotations and fMRI responses.

#### **Columns of the events table**

| Column | Meaning |
|---|---|
| `trial_type` | Identifies the controller action or game event, or marks a gameplay recording with `gym-retro_game`. |
| `onset` | Starting time in seconds from the beginning of the fMRI run. |
| `duration` | Duration in seconds. Instantaneous events have a duration of zero. |
| `stim_file` | Associated stimulus file, where applicable. For a `gym-retro_game` row, this identifies the gameplay recording. We use the filename to locate it. |
| `level` | Game level associated with the row, where provided. |
| `button` | Additional controller button information, where applicable. We use the `Action/...` labels in `trial_type` to identify controller actions. |
| `frame_start` | Starting frame index associated with the annotation. |
| `frame_stop` | Ending frame index associated with the annotation. |



Controller actions and game events may overlap in time. Their annotations retain their individual onsets and durations in seconds and are not grouped into the 1.49 second intervals (TR) used for fMRI sampling. We use these timestamps to align them with gameplay recordings and fMRI responses.

### **1.1.3 | Align gameplay and event annotations**

Each events table covers one **fMRI run**, a continuous recording of brain activity during which a subject can play several gameplay repetitions. Each repetition is saved as a separate **gameplay video**, containing both video and audio. One fMRI run can therefore span several gameplay videos.

The **beginning of the fMRI run is time zero for both the annotations and the corresponding fMRI data**. An annotation’s onset therefore indicates how many seconds have passed since that run began. Section 1.2 introduces the fMRI data, and Section 1.3 explains their alignment with gameplay.

Here, we focus on aligning the annotations with one selected gameplay video. The video player starts at **0 seconds at the beginning of that video**, which may begin partway through the fMRI run. Its `gym-retro_game` row gives the video’s starting time within the run, which we call **`video_start_in_run`**, and its duration.

For example, if the gameplay video starts **120 seconds into the fMRI run**, an action or game event beginning at run time **125.4 seconds** appears **5.4 seconds into the video**. We call the annotation’s starting time within the video **`event_onset_in_video`**. To calculate it, we subtract the video’s starting time within the run (`video_start_in_run`) from the annotation’s starting time within the run (`onset`):

```python
event_onset_in_video = onset - video_start_in_run
event_end_in_video = event_onset_in_video + duration
```

Here, `duration` is the annotation’s duration. If the annotation in this example lasts **0.3 seconds**, it ends at video time **5.7 seconds**.

The following functions load the matching annotations and convert their timestamps. We retain annotations that overlap the selected gameplay video, including sustained actions or events that began earlier but are still active when the video starts.

````python
def load_gameplay_events(video_path):
    """Return the matching TSV path, full run events and selected video's row."""

    # Get the video identifier from its filename.
    video_path = Path(video_path)
    video_id = video_path.stem.removesuffix('_recording')
    video_matches = []

    # Search the event tables stored in this video's session.
    for events_path in sorted(video_path.parent.glob('*_desc-annotated_events.tsv')):

        # Load the complete event table for this fMRI run.
        events = pd.read_csv(events_path, sep='\t', dtype={'level': str, 'stim_file': str})

        # Each gym-retro_game row describes one gameplay video,
        # including its filename, start time within the run and duration.
        gameplay_videos = events[events['trial_type'].eq('gym-retro_game')]

        # Compare each entry's stimulus filename with the selected video.
        for _, video_info in gameplay_videos.iterrows():
            stimulus_id = Path(str(video_info['stim_file'])).stem.removesuffix('_recording')

            # Store the matching TSV path, full run table and video row.
            if stimulus_id == video_id:
                video_matches.append((events_path, events, video_info))

    # The selected video must correspond to exactly one table row.
    if len(video_matches) != 1:
        raise ValueError(
            f'Expected one matching video entry for {video_path.name}; '
            f'found {len(video_matches)}.'
        )

    return video_matches[0]


def align_gameplay_events(events, video_info):
    """Select overlapping events and convert their times to the video timeline."""

    # Read timing from the selected video's gym-retro_game row.
    video_start_in_run = float(video_info['onset'])
    video_duration = float(video_info['duration'])

    # Calculate when the video and each event end on the fMRI run timeline.
    video_end_in_run = video_start_in_run + video_duration
    event_end_in_run = events['onset'] + events['duration']

    # Keep sustained events that overlap the video, including those
    # already active when the video starts.
    sustained = (
        (events['duration'] > 0)
        & (events['onset'] < video_end_in_run)
        & (event_end_in_run > video_start_in_run)
    )

    # Keep instantaneous events starting inside the video.
    # An event exactly at the video's end is excluded.
    instantaneous = (
        events['duration'].eq(0)
        & (events['onset'] >= video_start_in_run)
        & (events['onset'] < video_end_in_run)
    )

    # Select overlapping annotations, excluding rows describing videos.
    is_video_row = events['trial_type'].eq('gym-retro_game')
    aligned = events[(sustained | instantaneous) & ~is_video_row].copy()

    # Convert annotation times from the fMRI run timeline to the video timeline.
    aligned['event_onset_in_video'] = aligned['onset'] - video_start_in_run
    aligned['event_end_in_video'] = aligned['event_onset_in_video'] + aligned['duration']

    return aligned.sort_values('event_onset_in_video')
````

## **1.2 | fMRI data**

The Algonauts Project 2027 challenge uses fMRI data collected by [CNeuroMod](https://www.cneuromod.ca/) while subjects actively played video games. The released data include **parcel-wise fMRI responses** and **brain atlases**, which allow you to map these responses back into three-dimensional brain space.

This tutorial uses **`sub-01`**, who played all four games. In the full dataset, subject availability differs across games:

- **Mario, Mario All-Stars and Mario 3:** `sub-01`, `sub-02`, `sub-03`, `sub-05` and `sub-06`.
- **Shinobi:** `sub-01`, `sub-02`, `sub-04` and `sub-06`.

### **fMRI responses**

The fMRI responses are stored in `<game>/fmri/`, relative to the dataset root, with **one `.h5` file per subject and game**. These HDF5 files contain multiple arrays organized by scanning session and run.

The responses are provided in Montreal Neurological Institute (**MNI**) template space, identified by `MNI152NLin2009cAsym` in the filenames. Before parcel averaging, the voxel time series were corrected for motion, tissue and global signal contributions and slow drift, spatially smoothed with a **5 mm full width at half maximum** kernel, and standardized within each run. Voxel responses were then averaged within the **1,134 parcels of the CNeuroMod 2026 atlas**, combining cortical, subcortical and cerebellar regions.

Each run is stored as a two-dimensional array of shape **`(N samples, 1134 parcels)`**:

- **Rows** contain successive fMRI samples. The repetition time (**TR**) is **1.49 seconds**, so consecutive samples are acquired 1.49 seconds apart.
- **Columns** contain responses from the 1,134 brain parcels. A column refers to the same parcel throughout the run.

**One fMRI run can contain several gameplay recordings.** Its array therefore covers the continuous run, rather than one individual video. The corresponding event TSV uses the same run timeline, and its `gym-retro_game` rows identify each recording’s onset and duration.

#### **File naming convention**

The `.h5` files follow this naming convention:

`<subject>_task-<game>_space-MNI152NLin2009cAsym_atlas-cneuromod26_desc-1134Parcels_timeseries.h5`

The filename identifies the **subject**, **game**, **spatial template** and **atlas**. The `desc-1134Parcels` component indicates the number of parcel responses in each sample.

#### **Datasets within each file**

Each file contains session groups, with one dataset per fMRI run:

`ses-<session>/ses-<session>_task-<game>_run-<run>_timeseries`

- **`session`:** Scanning session identifier. Data were acquired across multiple visits to the scanner.
- **`game`:** Game played during the run.
- **`run`:** Continuous fMRI acquisition within that session.

For example, `ses-001/ses-001_task-mario_run-1_timeseries` contains the responses from Mario run 1 of session 001.


### **Brain atlases**

The brain atlases are stored in the shared `atlases/` directory and are independent of the game. Subject-specific atlases can be used to project a subject’s **1,134 parcel-wise values**, such as fMRI responses or Pearson’s $r$ scores, back into a 3D brain volume for plotting.

Each subject-specific atlas is a 3D array of shape **`(97, 115, 97)`**. Each voxel contains its parcel’s label, while `0` represents the background. The fMRI columns correspond to the **sorted unique positive labels in the subject’s atlas**.

The accompanying parcel metadata TSV provides anatomical names for the parcels.

## **1.3 | Align the fMRI responses with gameplay and events**

Here, you will align one gameplay recording from **`sub-01`** with its controller actions, annotated game events and corresponding fMRI responses. You will then use the subject’s brain atlas to map the **1,134 parcel responses** into three-dimensional brain space and visualize them alongside the gameplay.
### **Connect the video and event timestamps**

As discussed in Section 1.1.3, event onsets are measured from the **beginning of the run**. The fMRI responses follow the same timeline. An event with `onset = 125` therefore starts 125 seconds after the fMRI run began.

The video player instead shows how many seconds have passed since the **beginning of the gameplay recording**. The recording’s **`gym-retro_game` row** gives its onset within the run, allowing us to connect these two timelines.

The table below follows time downward. The fMRI and event timestamps advance together, while the video’s elapsed time begins when the recording starts.

| What happens | Time in the video | Time in the event table | Time in the fMRI run |
|---|---:|---:|---:|
| Run begins | Not started | 0 s | 0 s |
| Recording begins | 0 s | 120 s | 120 s |
| An event occurs | 5 s | 125 s | 125 s |
| Gameplay continues | 10 s | 130 s | 130 s |
To find the corresponding run time, we add the recording’s onset to the time shown in the video.

### **Find the corresponding fMRI sample**

The fMRI responses were acquired with a repetition time (**TR**) of **1.49 seconds**, meaning that one sample was acquired every 1.49 seconds.

To find the sample associated with a moment in the video, we first add the recording’s onset to obtain the corresponding run time, then divide by the TR and round down.

`sample index = floor((video time + recording onset) / TR)`

Here, `floor` rounds down to the nearest integer, and sample indices are **zero-based**.

In the table above, **5 seconds into the video** corresponds to **125 seconds in both the event and fMRI timelines**. The sample index is therefore `floor((5 + 120) / 1.49) = 83`. For this visualization, sample **83** represents the interval from **123.67 seconds up to, but not including, 125.16 seconds**.

Controller actions and game events appear according to their onsets and durations, while the brain map changes at the run’s TR boundaries.


### **Account for the delayed BOLD response**

<font color='red'><b>IMPORTANT:</b></font> fMRI detects the **BOLD (Blood Oxygen Level Dependent)** response, a signal that reflects changes in blood oxygenation associated with neural activity. When a brain region becomes active, local blood flow increases. This vascular response, described by the **hemodynamic response function (HRF)**, takes time and typically peaks around **5–6 seconds after a neural event**.

The `HRF_DELAY` parameter introduces a delay between the displayed gameplay and the selected fMRI sample, allowing a better correspondence between the gameplay and its associated brain response. It is expressed in **TRs**, rather than seconds.

`displayed sample index = acquisition sample index + HRF_DELAY`

For example, if the selected gameplay time corresponds to fMRI sample **17**, setting `HRF_DELAY = 3` displays sample **20**, acquired **4.47 seconds later**.

| fMRI sample index | Time relative to sample 17 | Role in this example |
|---|---:|---|
| **17** | **0 s** | Sample selected with `HRF_DELAY = 0` |
| 18 | +1.49 s | One TR later |
| 19 | +2.98 s | Two TRs later |
| **20** | **+4.47 s** | Sample selected with `HRF_DELAY = 3` |

The video and annotations remain at the selected gameplay time; only the displayed brain response shifts to a later sample.

### **View gameplay, annotations and brain responses together**

Below, you will load the matching fMRI run and subject-specific atlas, then define a function that maps parcel responses onto the brain.

The synchronized viewer displays **gameplay video and audio**, **controller actions and game events**, and **the corresponding brain responses**. Use `HRF_DELAY` to select how many TRs later the displayed response should be. Brain maps use the same colour scale throughout the recording so that responses can be compared.

````python
def plot_fmri_on_brain(fmri_run, atlas_img, sample_index, hrf_delay=0, figure=None, vmax=None):
    """
    Map one fMRI sample onto the subject atlas and plot a glass brain.

    Parameters
    ----------
    fmri_run : numpy.ndarray
        Run responses with shape (samples, parcels).
    atlas_img : nibabel.Nifti1Image
        Subject-specific atlas used to extract the parcel responses.
    sample_index : int
        Zero-based acquisition sample index before applying the delay.
    hrf_delay : int
        Number of TRs to shift forward when selecting the response.
    figure : matplotlib.figure.Figure or None
        Optional figure in which to draw the brain map.
    vmax : float or None
        Colour limit. Responses are shown on a scale from -vmax to +vmax.
        If omitted, the limit is determined from the selected sample.

    Returns
    -------
    brain_display : nilearn plotting display
        Brain plot, which can be saved or closed.
    """

    # Select the response sample, accounting for the hemodynamic delay.
    response_index = sample_index + hrf_delay
    if not 0 <= response_index < len(fmri_run):
        raise IndexError(f'fMRI sample {response_index} is outside the available run.')
    response = fmri_run[response_index]

    # Read the atlas and identify labels in fMRI column order.
    atlas_data = atlas_img.get_fdata().astype(np.int32)
    parcel_labels = np.unique(atlas_data[atlas_data > 0])
    if len(response) != len(parcel_labels):
        raise ValueError('The response length does not match the atlas parcel count.')

    # Assign each parcel response to all voxels carrying that label.
    brain_data = np.zeros(atlas_data.shape, dtype=np.float32)
    for column, label in enumerate(parcel_labels):
        brain_data[atlas_data == label] = response[column]

    # Preserve the atlas affine, which locates voxels in brain space.
    brain_img = nib.Nifti1Image(brain_data, atlas_img.affine)

    # Display positive and negative responses using a symmetric colour scale.
    if vmax is None:
        vmax = max(float(np.max(np.abs(response))), 1e-6)

    return plotting.plot_glass_brain(
        brain_img, figure=figure, display_mode='lyrz', cmap='RdBu_r',
        plot_abs=False, threshold=None, colorbar=True,
        symmetric_cbar=True, vmin=-vmax, vmax=vmax,
        title=f'fMRI sample {response_index}',
    )
````

## **2.1 | Feature extraction**

<font color='red'><b>Note:</b></font> Each modality (visual, audio, controller actions and semantic game events) requires different feature extraction steps. All examples use the same gameplay recording.

### **2.1.1 | Visual feature extraction**

Here, you will extract visual features from the gameplay video using [**Slow R50**](https://pytorch.org/hub/facebookresearch_pytorchvideo_resnet/), a convolutional neural network pretrained for action recognition on [**Kinetics 400**](https://arxiv.org/abs/1705.06950). The visual features consist of model activations produced by gameplay frames given as input.

Before passing the frames to the model, we apply the following preprocessing:

- **`UniformTemporalSubsample`:** uniformly selects 8 frames across each clip. This gives the model the same number of input frames for every TR.
- **`Lambda`:** divides pixel values by 255, converting their range from 0–255 to 0–1.
- **`Normalize`:** centers and scales each RGB channel using a mean of 0.45 and a standard deviation of 0.225.
- **`ShortSideScale`:** resizes the shorter image side to 256 pixels while preserving the aspect ratio.
- **`CenterCrop`:** selects a central region of 256 × 256 pixels.

<font color='red'><b>NOTE:</b></font> These preprocessing steps are specific to this Slow R50 model. Other models may require different frame counts, image sizes or normalization values.

You will begin by defining the frame transformation.

````python
def define_frames_transform():
    """
    Define the frame preprocessing used by Slow R50.

    Returns
    -------
    transform : torchvision.transforms.Compose
        Converts a tensor of shape (channels, frames, height, width) into
        a normalized clip containing 8 frames of size 256 × 256 pixels.
    """

    transform = Compose([
        # Select 8 frames uniformly across the clip.
        UniformTemporalSubsample(8),

        # Scale pixel values to the range 0–1.
        Lambda(lambda frames: frames / 255.0),

        # Normalize each RGB channel.
        Normalize([0.45, 0.45, 0.45], [0.225, 0.225, 0.225]),

        # Resize the shorter side while preserving the aspect ratio.
        ShortSideScale(size=256),

        # Crop the center of each frame.
        CenterCrop(256),
    ])

    return transform


# Define the frame transformation.
transform = define_frames_transform()
````

Before extracting the features, we need to know **when the recording began within its fMRI run and how long it lasted**. This allows us to select **chunks of 1.49 seconds**, matching the fMRI repetition time (**TR**). We define these chunks from the beginning of the run and keep only those that fall entirely within the recording.

The following function finds the recording’s **`gym-retro_game` row** in its session’s events tables and reads its onset and duration. It matches the recording using the filename in `stim_file`.

The onset is measured in seconds from the beginning of the run. We will use the same timing information to select matching chunks for visual and audio feature extraction.

````python
def load_recording_timing(video_path):
    """
    Load a gameplay recording's onset and duration within its fMRI run.

    Parameters
    ----------
    video_path : str or pathlib.Path
        Path to the gameplay recording.

    Returns
    -------
    recording_onset : float
        Recording start time in seconds from the beginning of the run.
    recording_duration : float
        Recording duration in seconds.
    """

    # Identify the recording from its video filename.
    video_path = Path(video_path)
    recording_id = video_path.stem.removesuffix('_recording')

    # Search the event tables stored in the recording's session.
    for events_path in sorted(video_path.parent.glob('*_desc-annotated_events.tsv')):
        events = pd.read_csv(events_path, sep='\t')

        # Each gym-retro_game row describes one recording, including
        # its filename, start time within the run and duration.
        recordings = events[events['trial_type'] == 'gym-retro_game']

        for recording in recordings.itertuples():

            # Match the stimulus filename to the selected recording.
            stimulus_id = Path(recording.stim_file).stem.removesuffix('_recording')
            if stimulus_id == recording_id:

                # Read this recording's timing on the fMRI run timeline.
                recording_onset = float(recording.onset)
                recording_duration = float(recording.duration)

                return recording_onset, recording_duration

    raise ValueError(
        f'No matching recording found in the event tables for {video_path.name}.'
    )
````

The actual feature extraction works in three steps:

1. **Divide the recording into clips:** select one clip per complete TR of 1.49 seconds, using the TR boundaries measured from the beginning of the fMRI run.
2. **Preprocess each clip:** apply the frame transformation to select 8 frames, normalize their pixel values and prepare their spatial dimensions.
3. **Extract features:** pass the preprocessed clip through the model and flatten the selected activations into one feature vector.

To identify the complete TRs covered by a recording, we use its onset and duration from the `gym-retro_game` row:

* **`first_tr = ceil(onset / TR)`:** the index of the first TR beginning at or after the recording onset.
* **`last_tr = floor((onset + duration) / TR)`:** the index immediately after the last complete TR. This value is exclusive, so the selection stops before `last_tr`.

For example, a recording beginning at run time **2 seconds** first contains a complete TR starting at **2.98 seconds**, with TR index **2**. Within the video, this clip begins at **0.98 seconds**, calculated by subtracting the recording onset.

````python
def extract_visual_features(video_path, onset, duration, tr, feature_extractor, model_layer, transform, device):
    """
    Extract one visual feature vector per complete run TR.

    Parameters
    ----------
    video_path : str or pathlib.Path
        Path to the gameplay recording.
    onset : float
        Recording onset in seconds from the beginning of the fMRI run.
    duration : float
        Recording duration in seconds, taken from its boundary annotation.
    tr : float
        fMRI repetition time in seconds.
    feature_extractor : torch.nn.Module
        Pretrained model returning the selected layer's activations.
    model_layer : str
        Name of the layer from which visual features are extracted.
    transform : torchvision.transforms.Compose
        Frame preprocessing that produces 8 normalized frames per clip.
    device : str or torch.device
        Device on which the model runs ('cpu' or 'cuda').

    Returns
    -------
    visual_features : numpy.ndarray
        Float32 array of shape (TR, visual features).
        Row i represents run TR ceil(onset / tr) + i.
    """
    import av

    # Load the RGB video frames and their presentation timestamps.
    frames, times = [], []
    with av.open(str(video_path)) as container:
        stream = container.streams.video[0]
        stream.thread_type = 'AUTO'
        for frame in container.decode(stream):
            frames.append(frame.to_ndarray(format='rgb24'))
            times.append(float(frame.pts * stream.time_base))

    # Convert the decoded frames and timestamps into arrays.
    frames = np.stack(frames)
    times = np.asarray(times)

    # Find the complete TRs covered by the recording.
    first_tr = int(np.ceil(onset / tr))
    last_tr = int(np.floor((onset + duration) / tr))
    if last_tr <= first_tr:
        raise ValueError('The recording contains no complete TRs.')

    # Collect one feature vector per interval.
    visual_features = []

    # Loop over TRs on the fMRI run timeline.
    for run_tr in tqdm(range(first_tr, last_tr), desc='Visual features'):

        # Convert the TR boundaries to video time.
        start = run_tr * tr - onset
        end = (run_tr + 1) * tr - onset

        # Select frames whose timestamps fall within this TR.
        chunk_frames = frames[(times >= start) & (times < end)]
        if len(chunk_frames) == 0:
            raise ValueError(f'No frames found for run TR {run_tr}.')

        # Arrange frames as (channels, frames, height, width).
        inputs = torch.from_numpy(chunk_frames).permute(3, 0, 1, 2).float()

        # Preprocess the frames, add a batch dimension and move to the device.
        inputs = transform(inputs).unsqueeze(0).to(device)

        # Extract the selected layer's activations.
        with torch.no_grad():
            predictions = feature_extractor(inputs)
        output = predictions[model_layer]

        # Inspect the model output for the first clip.
        if run_tr == first_tr:
            print(f'{model_layer} output shape: {tuple(output.shape)}')

        # Flatten the activations into one feature vector.
        visual_features.append(output.flatten().cpu().numpy())

    # Return the feature matrix without saving it to disk.
    return np.asarray(visual_features, dtype=np.float32)
````

### **2.1.2 | Audio feature extraction**

Here, you will extract audio features from the gameplay recording using [**Mel-frequency cepstral coefficients (MFCCs)**](https://librosa.org/doc/main/generated/librosa.feature.mfcc.html). MFCCs provide a compact representation of a sound’s spectral characteristics, describing how its energy is distributed across frequencies.

You will extract the features in three steps:

1. **Load the audio:** read the recording’s soundtrack, average its channels to obtain a mono waveform, and resample it to **22,050 Hz**.
2. **Divide the audio into chunks:** select one chunk per complete **TR of 1.49 seconds**, matching the TRs used for visual extraction.
3. **Extract features:** calculate **20 MFCCs** within each chunk and average each coefficient over time, producing one audio feature vector per TR.

````python
def extract_audio_features(video_path, onset, duration, tr, sr, n_mfcc):
    """
    Extract audio features using Mel-frequency cepstral coefficients.

    Parameters
    ----------
    video_path : str or pathlib.Path
        Path to the gameplay recording.
    onset : float
        Recording onset in seconds from the beginning of the fMRI run.
    duration : float
        Recording duration in seconds, taken from its boundary annotation.
    tr : float
        fMRI repetition time in seconds.
    sr : int
        Audio sampling rate used to compute the features.
    n_mfcc : int
        Number of MFCC coefficients retained per TR.

    Returns
    -------
    audio_features : numpy.ndarray
        Float32 array of shape (complete TRs, MFCC coefficients).
        Row i represents run TR ceil(onset / tr) + i.
    """

    # Open the recording's soundtrack at 44,100 Hz.
    clip = AudioFileClip(str(video_path), fps=44100)
    try:
        # Read successive audio blocks and join them into one waveform.
        audio_chunks = list(clip.iter_chunks(chunksize=50000, fps=44100, quantize=False, logger=None))
        waveform = np.concatenate(audio_chunks, axis=0)
    finally:
        # Close the audio reader after loading.
        clip.close()

    # Average the channels to obtain mono audio.
    if waveform.ndim > 1:
        waveform = waveform.mean(axis=1)

    # Resample the waveform to the requested sampling rate.
    waveform = librosa.resample(waveform, orig_sr=44100, target_sr=sr)

    # Find the complete TRs covered by the recording.
    first_tr = int(np.ceil(onset / tr))
    last_tr = int(np.floor((onset + duration) / tr))
    if last_tr <= first_tr:
        raise ValueError('The recording contains no complete TRs.')

    # Collect one feature vector per TR.
    audio_features = []

    # Loop over complete TRs on the fMRI run timeline.
    for run_tr in tqdm(range(first_tr, last_tr), desc='Audio features'):

        # Convert the TR boundaries to video time.
        start = max(0.0, run_tr * tr - onset)
        end = min(duration, (run_tr + 1) * tr - onset)

        # Convert seconds to waveform indices and select the audio chunk.
        chunk = waveform[int(start * sr):int(end * sr)]
        if len(chunk) == 0:
            raise ValueError(f'No audio samples found for run TR {run_tr}.')

        # Compute MFCCs across the chunk's short analysis frames.
        mfcc = librosa.feature.mfcc(y=chunk, sr=sr, n_mfcc=n_mfcc)

        # Average each coefficient over time to obtain one feature vector.
        audio_features.append(mfcc.mean(axis=1))

    # Return the feature matrix without saving it to disk.
    return np.asarray(audio_features, dtype=np.float32)
````

### **2.1.3 | Controller action feature extraction**

Controller annotations tell us which actions the player performed and how long they lasted. To use this information in an encoding model, we need to convert it into numerical features. We therefore turn them into a matrix with **one row per TR and one column per action**, where each TR lasts **1.49 seconds**.

For sustained actions, each value is the **fraction of the TR during which the action was active**. For example, holding Right for **0.745 seconds**, half of a TR, gives a value of **0.5**. These fractions are approximated by marking activity every **0.01 seconds (100 Hz)**.

Actions lasting **less than 0.001 seconds**, including zero-duration presses, instead receive **1 in the TR containing their onset**. Here, 1 indicates an occurrence rather than activity throughout the TR.

A matrix containing three example action columns might look like this:

| TR | Jump | Right | Run |
|---|---:|---:|---:|
| 1 | 0.5 | 1.0 | 0.0 |
| 2 | 0.0 | 0.5 | 0.8 |
| 3 | 0.0 | 0.0 | 0.0 |

Several actions can be active simultaneously, so rows do not need to sum to one.

We use **nine shared controller action columns**, sorted alphabetically. Actions absent from a recording have zero values. First, we load these labels from `tutorial_files/event_vocabulary.json`.

<font color='red'><b>Note:</b></font> This is a baseline representation. We encourage you to explore other ways to represent brief presses and sustained activity.

````python
def load_action_vocabulary(vocabulary_path):
    """
    Load the ordered controller-action labels across games.

    Parameters
    ----------
    vocabulary_path : str or pathlib.Path
        JSON file containing action and event descriptions for each game.

    Returns
    -------
    action_types : list of str
        Unique controller-action labels in alphabetical column order.
    """

    # Load the action and event dictionaries.
    with open(vocabulary_path) as file:
        vocabulary = json.load(file)

    # Collect action labels across games.
    action_labels = []
    for game_data in vocabulary.values():
        action_labels.extend(game_data['actions'])

    # Keep one column per action in a consistent order.
    return sorted(set(action_labels))


# Load the shared controller-action vocabulary.
action_types = load_action_vocabulary(tutorial_files_dir / 'event_vocabulary.json')

# Inspect the feature-column order.
print(f'Number of controller actions: {len(action_types)}')
for column, action in enumerate(action_types):
    print(f'{column}: {action}')
````

To extract controller-action features, we also need the **event annotations for the recording's run**. These tell us which actions occurred, when they began and how long they lasted.

The following function finds the event table containing the selected recording. It returns the complete run annotations together with the recording's onset and duration. The extraction function will then select the actions belonging to that recording.

We will also use this loading function for semantic game-event extraction, since actions and game events are stored in the same table.

````python
def load_recording_annotations(video_path):
    """
    Load the run annotations and timing for a gameplay recording.

    Parameters
    ----------
    video_path : str or pathlib.Path
        Path to the gameplay recording.

    Returns
    -------
    run_events : pandas.DataFrame
        Event annotations for the recording's complete fMRI run.
    onset : float
        Recording onset in seconds from the beginning of the run.
    duration : float
        Recording duration in seconds.
    """

    # Identify the recording from its video filename.
    video_path = Path(video_path)
    recording_id = video_path.stem.removesuffix('_recording')

    # Search the event tables stored in the recording's session.
    for events_path in sorted(video_path.parent.glob('*_desc-annotated_events.tsv')):
        run_events = pd.read_csv(events_path, sep='\t', dtype={'level': str})

        # Select the rows describing gameplay recording boundaries.
        boundaries = run_events[run_events['trial_type'] == 'gym-retro_game']

        for boundary in boundaries.itertuples():

            # Match the stimulus filename to the selected recording.
            stimulus_id = Path(boundary.stim_file).stem.removesuffix('_recording')
            if stimulus_id == recording_id:
                return run_events, float(boundary.onset), float(boundary.duration)

    raise ValueError(f'No recording boundary found for {video_path.name}.')
````

You will extract the features in three steps.

1. **Select the annotations:** keep the listed actions whose onsets fall within the recording.
2. **Mark active periods:** represent sustained actions as active or inactive on a timeline with one position every **0.01 seconds**. Overlapping annotations for the same action mark the same positions and are not counted twice.
3. **Average within each TR:** calculate the fraction of active positions in each complete **TR of 1.49 seconds**.

Actions lasting less than **0.001 seconds**, including zero-duration presses, receive **1** directly in the TR containing their onset. If the same action also has sustained activity in that TR, we take the maximum of the two values rather than adding them.

We use the same complete TRs as for the visual and audio features. Annotations beginning before the recording are excluded.

````python
def extract_event_features(run_events, onset, duration, event_types, tr, sampling_rate=100):
    """
    Convert event annotations into activity features per complete TR.

    Parameters
    ----------
    run_events : pandas.DataFrame
        Event annotations for the recording's fMRI run.
    onset : float
        Recording onset in seconds from the beginning of the run.
    duration : float
        Recording duration in seconds.
    event_types : list of str
        Event labels in feature-column order.
    tr : float
        fMRI repetition time in seconds.
    sampling_rate : int
        Activity timeline positions per second, with a default of 100.

    Returns
    -------
    event_features : numpy.ndarray
        Float32 array of shape (complete TRs, event labels).
        Sustained events receive activity fractions. Events lasting less
        than 0.001 seconds receive 1 in their onset TR.
        Row i describes run TR ceil(onset / tr) + i.
    """

    # Find the complete TRs covered by the recording.
    first_tr = int(np.ceil(onset / tr))
    last_tr = int(np.floor((onset + duration) / tr))
    n_trs = last_tr - first_tr

    if n_trs <= 0:
        raise ValueError('The recording contains no complete TRs.')

    # Select the requested events starting within the recording.
    events = run_events[
        run_events['trial_type'].isin(event_types)
        & (run_events['onset'] >= onset)
        & (run_events['onset'] < onset + duration)
    ]

    # Assign each event label to its feature column.
    columns = {label: column for column, label in enumerate(event_types)}

    # Create a timeline for sustained activity and a matrix for brief events.
    n_samples = int(duration * sampling_rate)
    activity = np.zeros((len(event_types), n_samples), dtype=bool)
    occurrences = np.zeros((n_trs, len(event_types)), dtype=np.float32)

    for event in events.itertuples():
        column = columns[event.trial_type]

        # Record near-zero-duration events directly in their onset TR.
        if event.duration < 0.001:
            row = int(np.floor(event.onset / tr)) - first_tr
            if 0 <= row < n_trs:
                occurrences[row, column] = 1.0

        else:
            # Convert sustained event times to recording-relative positions.
            start = int((event.onset - onset) * sampling_rate)
            end = min(int((event.onset - onset + event.duration) * sampling_rate), n_samples)

            # Combine overlapping annotations without double-counting.
            activity[column, start:end] = True

    # Calculate sustained activity fractions within complete TRs.
    event_features = np.zeros((n_trs, len(event_types)), dtype=np.float32)

    for row, run_tr in enumerate(range(first_tr, last_tr)):

        # Convert TR boundaries to recording-relative positions.
        start = int(max(0.0, run_tr * tr - onset) * sampling_rate)
        end = min(int(min(duration, (run_tr + 1) * tr - onset) * sampling_rate), n_samples)

        if end > start:
            event_features[row] = activity[:, start:end].mean(axis=1)

    # Use 1 where a brief event occurred, otherwise retain the activity fraction.
    return np.maximum(event_features, occurrences)
````

### **2.1.4 | Semantic game event feature extraction**

Game event annotations describe what happened during gameplay, such as collecting an item, defeating an enemy or taking damage. Different games can use different event labels, so we want a **shared representation that captures their meaning**.

We use **`all-MiniLM-L6-v2`**, a pretrained sentence encoder, to convert each event’s **description** into a numerical vector. This encoder produces **384 values per description**. Descriptions with similar meanings tend to produce related vectors. For example, collecting a coin in Mario and collecting an item in another game may have related representations, although semantic similarity does not guarantee similar brain responses.

For each **TR**, we calculate event activity using the same approach as for controller actions in Section 2.1.3. Sustained events receive the **fraction of the TR during which they were active**. Events lasting less than **0.001 seconds** receive **1 in the TR containing their onset**. Absent events receive 0.

We then **multiply each event’s vector by its activity value and add the resulting vectors**. For example, an enemy visible for half a TR contributes half its vector, while an instantaneous coin collection contributes its full vector.

Finally, we append the **sum of the event activity values** as one additional column. In this example, that value is **0.5 + 1 = 1.5**, so it represents total activity rather than a literal event count. The resulting feature matrix contains **one row per TR and 385 columns: 384 semantic dimensions plus one total activity column**.

First, we load the event labels and descriptions from `tutorial_files/event_vocabulary.json`.

<font color='red'><b>Note:</b></font> This is a baseline representation. We encourage you to explore alternative ways to represent game events.

````python
def load_game_event_vocabulary(vocabulary_path):
    """
    Load the ordered game-event labels and their descriptions.

    Parameters
    ----------
    vocabulary_path : str or pathlib.Path
        JSON file containing action and event descriptions for each game.

    Returns
    -------
    game_event_types : list of str
        Unique game-event labels in alphabetical feature-column order.
    event_descriptions : dict
        Text description associated with each event label.
    """

    # Load the action and event dictionaries.
    with open(vocabulary_path) as file:
        vocabulary = json.load(file)

    # Combine event descriptions across games.
    event_descriptions = {}
    for game_data in vocabulary.values():
        event_descriptions.update(game_data['events'])

    # Define the order used for activity columns and embedding rows.
    return sorted(event_descriptions), event_descriptions


# Load the shared game-event labels and descriptions.
vocabulary_path = tutorial_files_dir / 'event_vocabulary.json'
game_event_types, event_descriptions = load_game_event_vocabulary(vocabulary_path)

# Inspect one example description.
example_event_label = game_event_types[0]
print(f'Loaded {len(game_event_types)} game-event labels.')
print(f'\n{example_event_label}')
print(event_descriptions[example_event_label])
````

#### **Load the language model**

We use a fixed model revision so that the same encoder is loaded whenever the tutorial is run.

You will now encode the event descriptions into a matrix with dimensions **`(Event labels × 384 semantic dimensions)`**. Each row contains one event label’s embedding.

The vocabulary function sorts event labels alphabetically to establish a consistent order. We use this same order for the **columns of the event activity matrix** and the **rows of the embedding matrix**, ensuring that each activity value multiplies the correct event vector. This ordering concerns event labels. The **rows of the event activity matrix represent successive TRs** and remain in chronological order.

After encoding, we **center and normalize the embeddings** by subtracting the average vector across the full vocabulary and normalizing each resulting vector to unit length. Centering expresses each event relative to the vocabulary average; normalization gives each event vector the same length before it is weighted by activity.

We call the resulting embedding matrix **`phi`**. We also retain the `embedding_mean`, so that descriptions of new events can later use the same centering.

````python
def extract_event_embeddings(event_descriptions, event_types, encoder):
    """
    Encode event descriptions, then center and normalize their vectors.

    Parameters
    ----------
    event_descriptions : dict
        Text description associated with each event label.
    event_types : list of str
        Event labels defining the embedding matrix's row order.
    encoder : sentence_transformers.SentenceTransformer
        Pretrained sentence encoder.

    Returns
    -------
    phi : numpy.ndarray
        Centered unit-length embeddings with shape (event labels, 384).
    embedding_mean : numpy.ndarray
        Mean of the original embeddings with shape (1, 384).
    """

    # Arrange descriptions in the event vocabulary's order.
    descriptions = [event_descriptions[label] for label in event_types]

    # Convert each description into a numerical vector.
    embeddings = encoder.encode(descriptions, convert_to_numpy=True, show_progress_bar=True)

    # Subtract the average embedding across the full vocabulary.
    embedding_mean = embeddings.mean(axis=0, keepdims=True)
    centered_embeddings = embeddings - embedding_mean

    # Calculate the length of each centered vector.
    vector_lengths = np.linalg.norm(centered_embeddings, axis=1, keepdims=True)
    if (vector_lengths == 0).any():
        raise ValueError('An event has a zero-length centered embedding.')

    # Normalize each vector to unit length.
    phi = centered_embeddings / vector_lengths

    return phi, embedding_mean


# Extract embeddings in the same order as the game-event labels.
phi, embedding_mean = extract_event_embeddings(event_descriptions, game_event_types, event_encoder)

# Inspect the matrix dimensions and one example embedding.
print('Event embedding matrix shape:', phi.shape)
print('(Event labels × Embedding dimensions)')
print(f'\nExample event: {example_event_label}')
print('First ten embedding values:', phi[0, :10])
````

#### **Combine event activity and meaning**

Each event label now has a semantic vector stored in `phi`. For each TR, we **multiply each event’s vector by its activity value and add the contributions**, using the activity values calculated by `extract_event_features`.

For example, an enemy visible for half a TR contributes half its vector, while an instantaneous coin collection contributes its full vector:

`tr_vector = 0.5 * enemy_vector + coin_vector`

We calculate this for all TRs in three steps:

1. **Calculate event activity** using `extract_event_features`. The resulting `event_features` matrix has one row per TR and one column per event label.
2. **Combine activity and meaning** using `semantic_features = event_features @ phi`. This multiplies each event vector by its activity value and sums the contributions within each TR.
3. **Append total activity** as the final column by summing the activity values across event labels. In the example above, this value is **1.5**. It provides the encoding model with an explicit measure of total annotated event activity.

The columns of `event_features` and rows of `phi` must follow the same event label order.

The result has **one row per TR and 385 columns: 384 semantic dimensions plus one total activity column**. Before scaling, a TR without events has zeros in all 385 columns. Section 2.2 demonstrates scaling; Section 3 loads features that are already prepared and scaled.

````python
def extract_semantic_features(run_events, onset, duration, event_types, phi, tr):
    """
    Combine game-event activity with semantic embeddings.

    Parameters
    ----------
    run_events : pandas.DataFrame
        Event annotations for the recording's fMRI run.
    onset : float
        Recording onset in seconds from the beginning of the run.
    duration : float
        Recording duration in seconds.
    event_types : list of str
        Game-event labels in feature-column order.
    phi : numpy.ndarray
        Event embeddings with rows in the same order as event_types.
    tr : float
        fMRI repetition time in seconds.

    Returns
    -------
    game_event_features : numpy.ndarray
        Float32 array with one row per complete TR interval and
        384 semantic dimensions followed by one activity column.
    """

    # Calculate activity using the shared event extraction function.
    event_features = extract_event_features(run_events, onset, duration, event_types, tr)

    # Weight event embeddings by activity and add their contributions.
    semantic_features = event_features @ phi

    # Calculate total event activity within each interval.
    total_activity = event_features.sum(axis=1, keepdims=True)

    # Append activity as the final feature column.
    game_event_features = np.concatenate([semantic_features, total_activity], axis=1)

    return game_event_features.astype(np.float32)
````

## **2.2 | Scale stimulus features and reduce their dimensionality**

Here, you will learn how to prepare stimulus features for encoding model training using **scaling** and **principal component analysis (PCA)**.

For video and audio, we first **standardize each feature column** by subtracting its mean and dividing by its standard deviation. This procedure, also called **z-scoring**, gives each varying column a mean of zero and a standard deviation of one. Without this step, columns with larger variances can dominate PCA simply because of their numerical scale.

We then apply **PCA**, which combines the standardized features into new columns called *principal components*, ordered by how much variance they explain. Retaining fewer components reduces the size of the feature matrices and the cost of encoding model training, while discarding some variation.

Finally, we **standardize the retained components** before ridge regression. Ridge regression penalizes large model weights. A feature with larger values needs a smaller weight to make the same contribution to a prediction, resulting in a smaller penalty. Standardizing the components makes this penalty comparable across components.

The challenge baseline model uses the following preparation for each modality:

| Modality | Preparation |
|---|---|
| **Video** | Standardize the original columns → retain 100 PCA components → standardize the components |
| **Audio** | Standardize the original columns → retain 10 PCA components → standardize the components |
| **Controller actions** | Retain all nine columns and divide them by one shared scale |
| **Semantic game events** | Retain all 384 embedding columns and divide them by one shared scale; standardize the activity column separately |

Controller actions and semantic game events do not undergo PCA. Using one shared scale preserves the relative magnitudes within each block without separately amplifying columns with very little variation.

The examples below illustrate these steps using supplied **raw, unreduced and unscaled features** from the same Mario recording used in Section 2.1:

```text
stimulus_features/raw/sub-01/mario/<modality>/features.npy
```

For the supplied features used in Section 3, these quantities are calculated from training recordings only and then applied unchanged to both training and testing recordings. Section 3 loads the resulting features, so these steps do not need to be repeated there.

The following function standardizes each feature column by subtracting its mean and dividing by its standard deviation.

We use this function **before PCA** to give the original columns comparable scales. Later, we also use it **after PCA** to standardize the retained components before encoding model training.

````python
def standardize_features(features):
    """
    Standardize each feature column to mean zero and unit variance.

    Parameters
    ----------
    features : numpy.ndarray
        Feature matrix with shape (samples, features).

    Returns
    -------
    features_scaled : numpy.ndarray
        Standardized feature matrix with unchanged dimensions.
    """

    # Define the feature scaler.
    scaler = StandardScaler()

    # Estimate column means and standard deviations, then scale the features.
    features_scaled = scaler.fit_transform(features)

    return features_scaled
````

Next, you will define the PCA function.

The parameter **`n_components`** determines how many principal components to retain. More components preserve more variance, while fewer components produce a smaller representation.

We apply PCA to features whose columns have already been standardized.

For this demonstration, the component count must not exceed the original feature dimension and must be smaller than the number of samples. The function also reports the **explained variance**: the fraction of variation in the standardized features retained by the selected components.

````python
def perform_pca(features, n_components):
    """
    Reduce feature dimensionality using principal component analysis.

    Parameters
    ----------
    features : numpy.ndarray
        Feature matrix with shape (samples, features).
    n_components : int
        Number of principal components to retain.

    Returns
    -------
    features_pca : numpy.ndarray
        Reduced matrix with shape (samples, principal components).
    """

    # Determine the component limit for this centered-data demonstration.
    max_components = min(features.shape[0] - 1, features.shape[1])
    if not 1 <= n_components <= max_components:
        raise ValueError(f'Choose between 1 and {max_components} components.')

    # Define PCA with the requested number of components.
    pca = PCA(n_components=n_components, random_state=42)

    # Fit PCA and project the features onto the retained components.
    features_pca = pca.fit_transform(features)

    # Print the reduced dimensions and retained variance.
    print(f'\nPCA features shape: {features_pca.shape}')
    print('(Complete TR intervals × Principal components)')
    print(f'Explained variance: {pca.explained_variance_ratio_.sum():.1%}')

    return features_pca
````

#### **Scale the features used for encoding**

We now demonstrate the preparation used for the supplied encoding features.

| Modality | Reduction | Scaling |
|---|---|---|
| Video | 100 PCA components | Standardize each component |
| Audio | 10 PCA components | Standardize each component |
| Controller actions | None: retain 9 columns | One shared scale |
| Semantic events | None: retain 385 columns | One shared scale for the embeddings; standardize activity separately |

A **shared scale** divides every column in a block by the same number, preserving relative action magnitudes and semantic geometry:

`scale = sqrt(sum of column variances / rank)`

Here, **rank** estimates how many independent directions vary in the centered example data. For semantic events, we calculate it from the 384 embedding columns, excluding activity.

The division does **not subtract the mean**, so absent actions and event-free embeddings remain zero. The separately standardized activity column can become nonzero.

Choose a modality below. The video and audio demonstrations use the small PCA component count selected above. Controller actions and semantic events retain their original columns.

````python
def scale_feature_block(features):
    """
    Scale a feature block using one shared divisor.

    Parameters
    ----------
    features : numpy.ndarray
        Feature matrix with shape (samples, features).

    Returns
    -------
    features_scaled : numpy.ndarray
        Uncentered, scaled matrix with unchanged dimensions.
    """

    # Center temporarily to measure independently varying directions.
    centered_features = features.astype(np.float64)
    centered_features -= centered_features.mean(axis=0)
    singular_values = np.linalg.svd(centered_features, compute_uv=False)
    rank = np.sum(singular_values > 1e-5 * singular_values[0])

    # Leave a block unchanged if it has no variation.
    if rank == 0:
        return features.copy()

    # Divide all columns by the same scale.
    total_variance = centered_features.var(axis=0).sum()
    scale = np.sqrt(total_variance / rank)

    return (features / scale).astype(np.float32)
````

- **`hrf_delay`:** Nonnegative integer introducing a delay between stimulus samples and fMRI samples. fMRI detects the BOLD (Blood Oxygen Level Dependent) response, a signal that reflects changes in blood oxygenation levels in response to activity in the brain. Blood flow increases to a given brain region in response to its activity. This vascular response, which follows the hemodynamic response function (HRF), takes time. Typically, the HRF peaks around 5–6 seconds after a neural event. This delay reflects the time needed for blood oxygenation changes to propagate and for the fMRI signal to capture them. For example, with a `hrf_delay` of 3, if the fMRI sample of interest is 20, the corresponding stimulus sample will be 17. Each TR lasts **1.49 seconds**, so a delay of 3 TRs corresponds to **4.47 seconds**.

- **`stimulus_window`:** Positive integer indicating how many stimulus feature samples are used to model each fMRI sample, starting from the stimulus sample corresponding to the fMRI sample of interest, minus the `hrf_delay`, and going back in time. For example, with a `stimulus_window` of **4** and a `hrf_delay` of **3**, if the fMRI sample of interest is 20, it will be modeled with stimulus samples **[14, 15, 16, 17]**. Their feature vectors are **concatenated from oldest to newest**, rather than averaged. This applies to every selected modality. A window of 4 therefore multiplies the number of input features by 4. Also note that a larger stimulus window will increase compute time, since it increases the amount of stimulus features used to train and test the fMRI encoding models.

- **`levels_train`:** List of strings indicating the games and levels used to train the encoding model. You can test on held-out levels within a training game, or select all levels of another game and exclude that game from training to evaluate cross-game predictions.

- **`alphas`:** List of candidate ridge-regression penalties, set to [`0.1`, `1`, `10`, `100`, `1000`, `10000`]. Ridge regression penalizes large model weights to help control overfitting. Larger alpha values apply a stronger penalty; smaller values allow the model to fit the training data more closely.

- **`cv_folds`:** Integer indicating the number of cross-validation folds used to select alpha. We divide the concatenated training samples into **five contiguous folds without shuffling**. For each candidate alpha, the model is trained on four folds and evaluated on the remaining fold. Each parcel receives the alpha with the highest **mean R² across folds**, after which its model is fitted on all training samples. The selected test levels do not participate in alpha selection.

With the default settings, this excludes the first **six response samples** of each recording. There is no additional trailing exclusion. The following function returns the features organized by modality and recording identifier:

`features[modality][recording_id] = feature_matrix`

````python
def load_stimulus_features(tutorial_data_dir, subject, modality):
    """
    Load prepared stimulus features for the selected subject and modality.

    Parameters
    ----------
    tutorial_data_dir : pathlib.Path
        Main tutorial folder containing stimulus_features.
    subject : str
        Subject identifier, such as 'sub-01'.
    modality : str
        Feature modality ('video', 'audio', 'controller_actions', 'game_events'
        or 'all').

    Returns
    -------
    features : dict
        Feature matrices organized by modality and recording identifier.
        Each matrix has shape (complete TRs, features).
    """

    features = {}
    feature_dimensions = {
        'video': 100,
        'audio': 10,
        'controller_actions': 9,
        'game_events': 385,
    }

    # Load each selected modality's recording dictionary.
    for feature_modality, n_features in feature_dimensions.items():
        if modality not in [feature_modality, 'all']:
            continue

        features_path = (
            tutorial_data_dir / 'stimulus_features' / 'prepared'
            / subject / feature_modality / 'features.npy'
        )
        recordings = np.load(features_path, allow_pickle=True).item()

        # Check the expected feature dimensions before combining modalities.
        for recording_id, values in recordings.items():
            if values.ndim != 2 or values.shape[1] != n_features:
                raise ValueError(
                    f'Unexpected {feature_modality} shape for {recording_id}: '
                    f'{values.shape}; expected (TRs, {n_features}).'
                )

        features[feature_modality] = recordings

    return features
````

### **3.2.2 | Load the fMRI responses**

Next, you will load the fMRI responses for the selected subject from each game's `<game>/fmri/` directory. We load every game included in either the training or testing selection.

Each HDF5 file contains responses organized by session and run. Each run has dimensions **`(Run samples × 1,134 parcels)`** and can contain several gameplay recordings.

The function returns a dictionary indexed by `(game, session, run)`. Run numbers are converted to integers so that `run-01` in an event filename and `run-1` in an HDF5 dataset identify the same run.

We retain the full run responses here. During alignment, recording onsets and durations determine which response samples belong to each recording.

````python
def load_fmri(root_data_dir, subject, games):
    """
    Load full-run fMRI responses for the selected subject and games.

    Parameters
    ----------
    root_data_dir : pathlib.Path
        Root directory of the gameplay and fMRI dataset.
    subject : str
        Subject identifier, such as 'sub-01'.
    games : list of str
        Games included in training or testing.

    Returns
    -------
    fmri : dict
        Responses indexed by (game, session, run).
        Each array has shape (run samples, 1134 parcels).
    """

    fmri = {}

    for game in games:
        # Locate the subject's fMRI file for this game.
        paths = sorted((root_data_dir / game / 'fmri').glob(
            f'{subject}_task-{game}_*desc-1134Parcels_timeseries.h5'
        ))
        if len(paths) != 1:
            raise FileNotFoundError(
                f'Expected one fMRI file for {subject} in {game}; '
                f'found {len(paths)}.'
            )

        # Load the responses from each session and run.
        with h5py.File(paths[0], 'r') as file:
            for session in sorted(file):
                for run_name in file[session]:
                    run = int(re.search(r'run-(\d+)', run_name).group(1))
                    responses = file[session][run_name][:].astype(np.float32)

                    # Prevent different dataset names from overwriting a run.
                    key = (game, session, run)
                    if key in fmri:
                        raise ValueError(f'Duplicate fMRI run: {key}.')

                    fmri[key] = responses

    return fmri
````

## **3.3 | Encoding model training**

Now that you have loaded the stimulus features and fMRI responses, you will use the levels specified in `levels_train` to train an encoding model.

We first select the training recordings, construct delayed stimulus windows and align them with the corresponding fMRI responses. We then fit a ridge-regression model to predict each parcel's responses.

### **3.3.1 | Align the stimulus features and fMRI response samples for the train set**

Here, you will select and align the stimulus features and fMRI responses for the specified training levels. The goal is to bring the responses to a **`(Train samples × Parcels)`** format and the features to a **`(Train samples × Features)`** format, with matching sample counts.

First, we identify each recording’s level, onset and duration from its events table. Recordings without available fMRI responses are skipped.

Next, we combine the selected modalities for each TR. For each fMRI response sample, we then concatenate the feature vectors from earlier TRs according to the chosen delay and stimulus window. With a delay of **3 TRs** and a window of **4 TRs**, response sample 20 uses feature samples **14, 15, 16 and 17**, concatenated from oldest to newest.

Windows remain within each recording. Responses without a complete window are excluded, and recordings extending beyond the scan are limited to the available fMRI. Finally, we combine the aligned samples across recordings.

````python
def align_features_and_fmri_samples(features, fmri, root_data_dir, subject,
                                   levels, hrf_delay, stimulus_window, tr):
    """
    Align stimulus features and fMRI responses for the selected levels.

    Recording boundaries identify the complete TRs belonging to each
    recording. Feature vectors from earlier TRs are then concatenated
    to predict each response, accounting for the BOLD delay.

    Parameters
    ----------
    features : dict
        Prepared features indexed by modality and recording identifier.
        Each matrix has shape (complete TRs, features).
    fmri : dict
        Full run responses indexed by (game, session, run).
        Each array has shape (run samples, 1134 parcels).
    root_data_dir : pathlib.Path
        Root dataset directory containing the games' events tables.
    subject : str
        Subject identifier, such as 'sub-01'.
    levels : list of str
        Game/level identifiers selected for training or testing,
        such as 'mario/w1l1'.
    hrf_delay : int
        Delay between stimulus and response samples, in TRs.
        With a delay of 3, response sample 20 corresponds to stimulus
        sample 17.
    stimulus_window : int
        Number of consecutive TRs whose feature vectors are concatenated
        per response. With a window of 4 and a delay of 3, response
        sample 20 uses stimulus samples 14, 15, 16 and 17, ordered
        oldest to newest. The same window applies to every modality.
    tr : float
        fMRI repetition time in seconds, set to 1.49.

    Returns
    -------
    aligned_features : numpy.ndarray
        Float32 matrix with shape (samples, stimulus_window * total_features).
        Modalities are ordered alphabetically within each TR, and TRs
        are concatenated from oldest to newest.
    aligned_fmri : numpy.ndarray
        Float32 matrix with shape (samples, 1134 parcels), with rows
        corresponding to the same samples as aligned_features.
    """

    ### Prepare the output lists and temporal settings ###
    aligned_features = []
    aligned_fmri = []
    modalities = sorted(features)

    # Earlier responses do not have a complete delayed stimulus window.
    first_response = hrf_delay + stimulus_window - 1
    selected_games = sorted({level.split('/')[0] for level in levels})

    ### Loop across the selected games and their event tables ###
    for game in selected_games:
        events_dir = root_data_dir / game / 'gamelogs' / subject

        for path in sorted(events_dir.glob('ses-*/*_desc-annotated_events.tsv')):
            # Match the event table to its fMRI session and run.
            session = path.parent.name
            run = int(re.search(r'run-(\d+)', path.name).group(1))
            run_key = (game, session, run)

            # Skip gameplay runs without released fMRI responses.
            if run_key not in fmri:
                continue

            # Each recording-boundary row provides a level, onset and duration.
            events = pd.read_csv(path, sep='\t', dtype={'level': str})
            recordings = events[events['trial_type'].eq('gym-retro_game')]

            ### Loop across recordings belonging to the selected levels ###
            for recording in recordings.itertuples():
                if f'{game}/{recording.level}' not in levels:
                    continue

                # Convert the replay filename to the feature recording ID.
                recording_id = f'{Path(recording.stim_file).stem}_recording'


                # Skip recordings without supplied features
                if recording_id not in features[modalities[0]]:
                    continue

                # Combine modalities side by side within each TR.
                recording_features = np.concatenate(
                    [features[mod][recording_id] for mod in modalities], axis=1,
                )

                # Select the recording's complete TRs on the run's TR grid.
                # last_tr is exclusive, matching NumPy's slice convention.
                first_tr = int(np.ceil(recording.onset / tr))
                last_tr = int(np.floor(
                    (recording.onset + recording.duration) / tr
                ))

                # The supplied features must cover exactly these complete TRs.
                if len(recording_features) != last_tr - first_tr:
                    raise ValueError(
                        f'{recording_id}: {len(recording_features)} feature rows, '
                        f'but the recording spans {last_tr - first_tr} complete TRs.'
                    )

                # Extract matching responses; slicing stops at the scan's end.
                responses = fmri[run_key][first_tr:last_tr]

                ### Build one feature window for each usable response ###
                for response_index in range(first_response, len(responses)):
                    # Find the earlier stimulus intervals for this response.
                    window_end = response_index - hrf_delay + 1
                    window_start = window_end - stimulus_window
                    window = recording_features[window_start:window_end]

                    # Flatten oldest to newest and append the matching response.
                    aligned_features.append(window.flatten())
                    aligned_fmri.append(responses[response_index])

    ### Combine samples from all selected recordings ###
    aligned_features = np.asarray(aligned_features, dtype=np.float32)
    aligned_fmri = np.asarray(aligned_fmri, dtype=np.float32)

    return aligned_features, aligned_fmri
````

````python
# Select the training recordings and align their delayed feature windows
# with the corresponding fMRI responses.
features_train, fmri_train = align_features_and_fmri_samples(
    features, fmri, root_data_dir, subject,
    levels=levels_train, hrf_delay=hrf_delay,
    stimulus_window=stimulus_window, tr=TR,
)

# Inspect the training arrays. Their sample counts should match,
# since each feature window is paired with one fMRI response.
print('Training fMRI responses shape:')
print(fmri_train.shape)
print('(Train samples × Parcels)')

print('\nTraining stimulus features shape:')
print(features_train.shape)
print('(Train samples × Features)')
````

### **3.3.2 | Train the encoding model**

In this step, you will train an encoding model to predict fMRI responses to gameplay. The model is based on **ridge regression**, which learns a linear mapping from stimulus features to each parcel’s responses while penalizing large model weights to help control overfitting.

We use **`RidgeCV`** to select the regularization penalty, **alpha**, using **five fold cross validation** on the training data. Each candidate alpha is evaluated using **R²**, averaged across parcels and folds. The alpha with the highest mean score is then used to fit the model on all training samples. Each parcel has its own regression weights, while the selected alpha is shared across all parcels.

````python
def train_encoding(features_train, fmri_train, alphas, cv_folds):
    """
    Train a ridge encoding model with one penalty shared across parcels.

    The penalty is selected using mean R² across parcels and folds.
    The selected value is available in model.alpha_.
    """

    # Use float64 for numerical precision.
    features_train = np.asarray(features_train, dtype=np.float64)
    fmri_train = np.asarray(fmri_train, dtype=np.float64)

    # Select the penalty using contiguous, unshuffled folds,
    # then fit the model on all training samples.
    model = RidgeCV(
        alphas=alphas,
        cv=KFold(n_splits=cv_folds, shuffle=False),
        scoring='r2',
    ).fit(features_train, fmri_train)

    return model
````

````python
# Select each parcel's ridge penalty and fit the encoding model.
model = train_encoding(features_train, fmri_train, alphas, cv_folds)

# Remove unused training arrays from memory.
del features_train, fmri_train
````

### **3.4.3 | Compute and plot the encoding accuracy**

Finally, you will compute and plot the encoding accuracy for the testing recordings, measured using **Pearson’s r**. The predicted and recorded fMRI responses are correlated independently at each parcel across all testing samples, resulting in one correlation score for each of the **1,134 parcels**.

We then average these scores across parcels to obtain the **mean Pearson’s r**. The individual parcel scores are mapped onto the subject’s **CNeuroMod 2026 atlas** and displayed on a glass brain.

````python
def compute_encoding_accuracy(fmri_test, fmri_test_pred, subject, modality):
    """
    Compute parcel correlations, print mean Pearson's r, and plot a glass brain.

    Parameters
    ----------
    fmri_test : ndarray of shape (n_samples, n_parcels)
        Recorded fMRI responses for the testing recordings.
    fmri_test_pred : ndarray of shape (n_samples, n_parcels)
        Predicted responses, in the same sample and parcel order.
    subject : str
        Subject identifier used to train and test the encoding model.
    modality : str
        Feature modality used to train and test the encoding model.

    Returns
    -------
    encoding_accuracy : ndarray of shape (n_parcels,)
        Pearson's r for each parcel.
    """

    ### Correlate recorded and predicted responses at each parcel ###
    encoding_accuracy = np.zeros(fmri_test.shape[1], dtype=np.float32)
    for parcel in range(len(encoding_accuracy)):
        encoding_accuracy[parcel] = pearsonr(
            fmri_test[:, parcel], fmri_test_pred[:, parcel]
        )[0]

    ### Print mean Pearson's r across parcels ###
    mean_pearson_r = np.mean(encoding_accuracy)
    print(f"Mean Pearson's r: {mean_pearson_r:.3f}")

    ### Map the parcel correlations onto the subject's brain atlas ###
    atlas_path = next(
        (root_data_dir / 'atlases').glob(
            f'{subject}_space-*_desc-1134Parcels_dseg.nii.gz'
        )
    )
    atlas_masker = NiftiLabelsMasker(labels_img=str(atlas_path))
    atlas_masker.fit()
    encoding_accuracy_nii = atlas_masker.inverse_transform(encoding_accuracy)

    ### Plot the parcel correlations ###
    title = (
        f'Encoding accuracy, {subject}, modality-{modality}, '
        f"mean Pearson's r: {mean_pearson_r:.3f}"
    )

    display = plotting.plot_glass_brain(
        encoding_accuracy_nii,
        display_mode='lyrz',
        cmap='hot_r',
        threshold=1e-6,
        colorbar=True,
        plot_abs=False,
        symmetric_cbar=False,
        title=title,
    )

    display._cbar.set_label(
        "Pearson's $r$", rotation=90, labelpad=12, fontsize=12
    )
    plotting.show()

    return encoding_accuracy
````
